```python
#question 1
import numpy as np

arr = np.random.randint(1, 101, size=(3, 3))

print("Original Array:")
print(arr)

min_val = arr.min()
max_val = arr.max()

normalized_arr = (arr - min_val) / (max_val - min_val)

print("\nNormalized Array:")
print(normalized_arr)
```

```python
#question 2
import numpy as np

arr = np.random.randint(1, 101, size=10)

print("Array:", arr)

x = float(input("Enter a value X: "))

closest = arr[np.abs(arr - x).argmin()]

print("Closest value:", closest)
```

```python
#question 3
import numpy as np

arr = np.random.randint(1, 101, size=(5, 5))

print("Original Array:")
print(arr)

sorted_arr = np.sort(arr, axis=0)

second_largest = sorted_arr[-2, :]

print("\nSecond-largest value in each column:")
print(second_largest)
```

```python
#question 4
import numpy as np

arr = np.random.randint(1, 101, size=(6, 4))

print("Original Array:")
print(arr)

row_sums = arr.sum(axis=1)

even_rows = np.where(row_sums % 2 == 0)[0]

print("Rows with even sums:", even_rows)
```

```python
#question 5
import numpy as np

arr = np.random.randint(1, 51, size=30)

print("Original Array:")
print(arr)

arr[(arr % 3 == 0) | (arr % 5 == 0)] = -1

print("\nModified Array:")
print(arr)
```

```python
#question 6
import numpy as np

arr = np.random.randint(1, 101, size=(3, 3))

print("Original Array:")
print(arr)

row_means = arr.mean(axis=1, keepdims=True)

result = arr - row_means

print("\nNormalized Array:")
print(result)
```
