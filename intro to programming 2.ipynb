```python
#question 1
def check_k(lst,k):
    new_lst=[]
    for i in lst:
        if len(i) > k :
            new_lst.append(i)
    return new_lst

lst=input("Enter the list: ").split(" ")
k= int(input("Enter the length: "))
print(check_k(lst,k))
```

```python
#question 2
def area(shape,r_or_l,w=0):
    pi=22/7
    if shape=='circle':
        area=pi*r_or_l**2
    else:
        area=r_or_l*w
    return area 

shape=input("Enter the shape: ")
if shape == "circle" :
    r=int(input("Enter the radius: "))
    print(area(shape,r))
else:
    l,w=map(int,input("Enter the length and width: ").split(" "))
    print(area(shape,l,w))
```

```python
#question 5

class Restaurant():
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer_name, table_number):
        reservation = {
            "customer": customer_name,
            "table": table_number
        }
        self.book_table.append(reservation)

    def customer_order(self, customer_name, item):
        order = {
            "customer": customer_name,
            "item": item
        }
        self.customer_orders.append(order)

    def print_menu(self):
        print("\n----- Restaurant Menu -----")
        for item, price in self.menu_items.items():
            print(f"{item}: ${price}")

    def print_reservations(self):
        print("\n----- Table Reservations -----")
        for reservation in self.book_table:
            print(f"Customer: {reservation['customer']}, "
                  f"Table: {reservation['table']}")

    def print_orders(self):
        print("\n----- Customer Orders -----")
        for order in self.customer_orders:
            print(f"Customer: {order['customer']}, "
                  f"Order: {order['item']}")


restaurant = Restaurant()

# Add items to the menu
restaurant.add_item_to_menu("Pizza", 12)
restaurant.add_item_to_menu("Burger", 8)
restaurant.add_item_to_menu("Pasta", 10)
restaurant.add_item_to_menu("Salad", 6)

# Make table reservations
restaurant.book_tables("Ahmed", 1)
restaurant.book_tables("Sara", 2)
restaurant.book_tables("Omar", 3)

# Take customer orders
restaurant.customer_order("Ahmed", "Pizza")
restaurant.customer_order("Sara", "Burger")
restaurant.customer_order("Omar", "Pasta")

restaurant.print_menu()

restaurant.print_reservations()

restaurant.print_orders()
```

```python
#question 6

class Customer():
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_age(self):
        return self.__age

    def set_age(self, age):
        self.__age = age



class Bike():
    def __init__(self, name, price):
        self.__name = name
        self.__price = price
    
    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_price(self):
        return self.__price

    def set_price(self, price):
        self.__price = price

        

class MainBikeRental:
    def __init__(self):
        self.__bikes = []
        self.__rentals = []

    def get_bikes(self):
        return self.__bikes

    def get_rentals(self):
        return self.__rentals

    def set_bikes(self, bikes):
        self.__bikes = bikes

    def set_rentals(self, rentals):
        self.__rentals = rentals

    def addBike(self, bike, quantity):
        for i in range(quantity):
            self.__bikes.append(bike)

    def totalCost(self):
        total = 0
        for bike in self.__bikes:
            total += bike.get_price()
        return total

    def requestBike(self, customer, number_of_bikes):

        if customer.get_age() <= 6:
            print("Customer must be older than 6 years.")
            return

        if number_of_bikes <= 0:
            print("Please enter a valid number of bikes.")
            return

        if number_of_bikes <= len(self.__bikes):

            rented_bikes = []

            for i in range(number_of_bikes):
                rented_bikes.append(self.__bikes.pop())

            rental = {
                "customer": customer.get_name(),
                "bikes": rented_bikes
            }

            self.__rentals.append(rental)

            print(f"{number_of_bikes} bike(s) rented successfully!")

        else:
            print("Not enough bikes available. Please wait 10 minutes.")

    def returnBike(self, rentalTime, num_of_bikes):

        if num_of_bikes <= 0 or num_of_bikes > len(self.__rentals):
            print("Invalid return request.")
            return

        rental = self.__rentals.pop(0)

        if num_of_bikes > len(rental["bikes"]):
            print("Invalid number of bikes.")
            self.__rentals.insert(0, rental)
            return

        returned_bikes = rental["bikes"][:num_of_bikes]

        rental["bikes"] = rental["bikes"][num_of_bikes:]

        self.__bikes.extend(returned_bikes)

        bill = rentalTime * 40

        print(f"Total bill: {bill}")
        print(f"{num_of_bikes} bike(s) returned successfully!")

        
```

```python
customer1 = Customer("Ahmed", 22)
customer2 = Customer("Omar", 5)

bike1 = Bike("Mountain Bike", 5000)
bike2 = Bike("Road Bike", 4000)
bike3 = Bike("Electric Bike", 10000)

shop = MainBikeRental()

shop.addBike(bike1, 3)
shop.addBike(bike2, 2)
shop.addBike(bike3, 1)

print("Total bikes value:", shop.totalCost())

shop.requestBike(customer1, 2)

shop.requestBike(customer2, 1)

shop.returnBike(3, 2)

print("Updated bikes value:", shop.totalCost())
```
