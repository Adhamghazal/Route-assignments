```python
#question 1
text=input("Enter a string consisting of letters and numbers: ")
total=0
word=''
for letter in text:
    if letter.isdigit():
        total+=int(letter)
    else:
        word+=letter
print("Word=",word,"sum=",total)
```

```python
#question 2
password=input("Enter your password: ")
contain_upper=False
contain_lower=False
contain_special=False
contain_digit=False
special_char= ['$', '@', '#', '%']

for char in password:
    if char in special_char:
        contain_special=True
    if char.isupper():
        contain_upper=True
    if char.islower():
        contain_lower=True
    if char.isdigit():
        contain_digit=True
if contain_upper and contain_lower and contain_special and contain_digit and len(password)>=8 and len(password)<=16:
    print("Password is Secure")
else:
    print("Password is inSecure")
```

```python
#question 8
nums=input("Enter the list: ").split(",")
nums=[int(i) for i in nums]
unique=set(nums)
result=[]
for i in unique:
    freq=0
    for j in nums:
        if i==j:
            freq+=1
    result.append((i,freq))
print(result)
```

```python
#question 9
num=int(input("enter a number: "))
sum_num=num*(num+1)//2
fact=1
for i in range(1,num+1):
    fact*=i
if fact%sum_num == 0:
    print("Factorial is",fact,"Sum is",sum_num,fact,"is divisble by",sum_num)
else:
    print("Factorial is",fact,"Sum is",sum_num,fact,"is not divisble by",sum_num)
```

```python
#question 10
market={
    'apple':0.5,
    'milk':1.5,
    'bread':2.0
}
total=0
process=input().split(",")
for i in process:
    start=i.find("'")+1
    end=i.find("'",start)
    key=i[start:end]
    key=key.replace(" ",'')
    i=i.replace("'"+key+"'",'')
    i=i.replace(" ",'')
    if key not in market:
        print("Item not available")
        continue
    if i == "Add" :
        total+=market[key]
    elif i =="Remove" and total == 0 :
        print("The cart is already empty")
    else:
        total-=market[key]
print(total)
```
